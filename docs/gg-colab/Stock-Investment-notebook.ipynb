## Models testings and playground

### Deepseek V4

```python
from google.colab import userdata
from openai import OpenAI

# Initialize DeepSeek client
deepseek_client = OpenAI(
    api_key=userdata.get('DEEPSEEK-V4-KEY'),
    base_url="https://api.deepseek.com"
)

response = deepseek_client.chat.completions.create(
    model="deepseek-v4-flash", # Use 'deepseek-chat' for V3 or 'deepseek-reasoner' for R1
    messages=[
        {"role": "user", "content": "Analyze the financial risks of HPG based on high leverage."}
    ]
)

# DeepSeek R1 returns reasoning content before the answer
if hasattr(response.choices[0].message, 'reasoning_content'):
    print("--- REASONING PROCESS ---")
    print(response.choices[0].message.reasoning_content)

print("\n--- FINAL ANSWER ---")
print(response.choices[0].message.content)
```
